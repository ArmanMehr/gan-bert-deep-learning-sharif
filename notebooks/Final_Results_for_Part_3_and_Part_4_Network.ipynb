# Using G1 of Part 3 and compare it to Part 1 of the Project with max_length=64

```
##### Read the data from Part 1 of the Project

workbook = openpyxl.load_workbook('data_for_maxlength64_epochs3.xlsx')
sheet = workbook['Sheet1']

percentage=np.zeros(4)
average_accs_for_all_percentages_for_Part1=np.zeros(4)
i=0
for row in sheet.iter_rows(values_only=True):
    if(i>0):
      average_accs_for_all_percentages_for_Part1[i-1]=row[1]
      percentage[i-1]=100*row[0]
    i=i+1

##### Read the data from Part 3 of the Project(G1 Network)


with open('Part3_stats_G1_1percent_maxlen64 .json', 'r') as f:
    data = json.load(f)
    acc_1=data[-1]["Valid. Accur."]

with open('Part3_stats_G1_5persent_maxlen64.json', 'r') as f:
    data = json.load(f)
    acc_5=data[-1]["Valid. Accur."]

with open('Part3_stats_G1_10persent_maxlen64.json', 'r') as f:
    # data = json.load(f)
    acc_10=data[-1]["Valid. Accur."]

with open('Part3_stats_G1_50persent_maxlen64.json', 'r') as f:
    data = json.load(f)
    acc_50=data[-1]["Valid. Accur."]

average_accs_for_all_percentages_for_Part3_G1_64=np.array([acc_1,acc_5,acc_10,acc_50])


plt.plot(percentage,average_accs_for_all_percentages_for_Part1, marker='o', linestyle='--',label='Part 1')
plt.plot(percentage,100*average_accs_for_all_percentages_for_Part3_G1_64, marker='o', linestyle='--',label='Part 3')
plt.title("Part 1 & 3 Accuracy, G1 , Max_length=64")
plt.ylabel("Test Accuracy")
plt.xlabel("percentage")
plt.legend()
plt.grid('on')
```
